```python
import pandas as pd
import ast
import numpy as np

full_train_csv = pd.read_csv("training.csv.gz")
full_test_csv = pd.read_csv("testing.csv.gz")
```

```python

# ============================================================
# EDA ON THE ENTIRE TRAIN DATASET
# ============================================================




# ============================================================
# duplicate shot_id checks
# ============================================================

print("\n---Duplicate checks--------")

missing_shot_ids = full_train_csv["shot_id"].isna().sum()
duplicate_shot_ids = full_train_csv["shot_id"].duplicated().sum()
duplicate_rows = full_train_csv.duplicated().sum()

print(f"Missing shot_id values: {missing_shot_ids:,}")
print(f"Duplicate shot_id values: {duplicate_shot_ids:,}")
print(f"Fully duplicated rows: {duplicate_rows:,}")

# ============================================================
# target-variable checks
# ============================================================

print("\n--- Outcome checks ---")

if "outcome" in full_train_csv.columns:
    print("Outcome counts, including missing values:")
    display(full_train_csv["outcome"].value_counts(dropna=False).to_frame())
    print(f"Missing outcomes: {full_train_csv['outcome'].isna().sum():,}")
    print(f"Overall make rate: {full_train_csv['outcome'].mean():.4f}")


# ============================================================
#  Missing-value summary
# ============================================================

missing_summary = pd.DataFrame({
    "missing_count": full_train_csv.isna().sum(),
    "missing_percent": full_train_csv.isna().mean().mul(100)
})

missing_summary["non_missing_count"] = (len(full_train_csv) - missing_summary["missing_count"])

missing_summary = missing_summary[
    ["missing_count", "missing_percent", "non_missing_count"]
].sort_values("missing_percent", ascending=False)

print("\nMissing-value summary:")
display(missing_summary)
# ============================================================
# unique vals for categorical columns
# ============================================================

categorical_columns = [
    "season_id",
    "off_team_id",
    "def_team_id",
    "shooter_id",
    "closestdef_id",
    "contester1_id",
    "contester2_id",
    "contester3_id",
    "contester4_id",
    "contested",
    "month",
    "gamestate",
    "shottype",
    "three"
]


categorical_summary = pd.DataFrame({
    "dtype": full_train_csv[categorical_columns].dtypes.astype(str),
    "unique_non_missing": full_train_csv[
        categorical_columns
    ].nunique(dropna=True),
    "missing_count": full_train_csv[
        categorical_columns
    ].isna().sum()
}).sort_values("unique_non_missing")

print("\nCategorical-column summary:")
display(categorical_summary)


# I'm diisplayign all values only for low-cardinality categorical cols

low_cardinality_columns = [
    "season_id",
    "contested",
    "month",
    "gamestate",
    "shottype",
    "three"
]

print("\nValues of low-cardinality categorical columns:")

for column in low_cardinality_columns:
    if column in full_train_csv.columns:
        values = (
            full_train_csv[column]
            .value_counts(dropna=False)
            .rename("count")
            .to_frame()
        )

        print(f"\n{column}:")
        display(values)


# ============================================================
# 8. Parse and validate closestdefapproach
# ============================================================
import ast
import re
import numpy as np
import pandas as pd


def parse_approach_dictionary(value):
    """
    Parse one closestdefapproach entry.

    Expected input:
        A string representation of a set containing four defender-distance measurements.
        Example: "{20.0470047638,18.8207810677,16.568961947,13.8624673129}"

    Returns:
        list  -> successfully parsed as list of 4 floats in original order
        None  -> missing
        np.nan -> malformed
    """

    if value is None or (isinstance(value, float) and pd.isna(value)):
        return None

    if not isinstance(value, str):
        value = str(value)

    #REGEX!!!!!!!
    pattern = r'-?\d+\.?\d*'
    matches = re.findall(pattern, value)
    
    if not matches:
        return np.nan
    
    try:
        
        numbers = [float(m) for m in matches]
        
       
        if len(numbers) != 4:
            return np.nan
        
        
        return numbers
    except ValueError:
        return np.nan


parsed_approach = full_train_csv[
    "closestdefapproach"
].apply(parse_approach_dictionary)
full_train_csv["closestdefapproach_parsed"] = parsed_approach

def is_missing_parsed_value(value):
    return value is None


def is_malformed_parsed_value(value):
    return isinstance(value, float) and np.isnan(value)


def has_four_entries(value):
    return isinstance(value, list) and len(value) == 4


def has_numeric_values(value):
    if not isinstance(value, list):
        return False

    return all(
        isinstance(
            item,
            (int, float, np.integer, np.floating)
        )
        and not pd.isna(item)
        for item in value
    )


approach_missing = parsed_approach.apply(
    is_missing_parsed_value
)

approach_malformed = parsed_approach.apply(
    is_malformed_parsed_value
)

approach_wrong_length = parsed_approach.apply(
    lambda value: (
        isinstance(value, list)
        and len(value) != 4
    )
)

approach_nonnumeric = parsed_approach.apply(
    lambda value: (
        isinstance(value, list)
        and not has_numeric_values(value)
    )
)

print("--- closestdefapproach validation ---")
print(f"Missing entries: {approach_missing.sum():,}")
print(f"Malformed entries: {approach_malformed.sum():,}")
print(f"Lists not containing four entries: {approach_wrong_length.sum():,}")
print(f"Lists containing nonnumeric values: {approach_nonnumeric.sum():,}")

# ============================================================
# 9. Impossible or suspicious numeric values
# ============================================================

nonnegative_columns = [
    "num_contesters",
    "distcont1",
    "distcont2",
    "distcont3",
    "distcont4",
    "distance",
    "dribblesbefore",
    "shotclock",
    "closestdefdist",
    "shooterspeed"
]

nonnegative_columns = [
    column for column in nonnegative_columns
    if column in full_train_csv.columns
]

range_checks = []

for column in nonnegative_columns:
    negative_count = full_train_csv[column].lt(0).sum()

    range_checks.append({
        "check": f"{column} is nonnegative",
        "violation_count": negative_count
    })

if "shotclock" in full_train_csv.columns:
    range_checks.append({
        "check": "shotclock <= 24",
        "violation_count": full_train_csv["shotclock"].gt(24).sum()
    })

if "num_contesters" in full_train_csv.columns:
    range_checks.append({
        "check": "num_contesters <= 4",
        "violation_count": full_train_csv[
            "num_contesters"
        ].gt(4).sum()
    })

if "locationx" in full_train_csv.columns:
    range_checks.append({
        "check": "-47 <= locationx <= 47",
        "violation_count": (
            full_train_csv["locationx"].lt(-47)
            | full_train_csv["locationx"].gt(47)
        ).sum()
    })

if "locationy" in full_train_csv.columns:
    range_checks.append({
        "check": "-25 <= locationy <= 25",
        "violation_count": (
            full_train_csv["locationy"].lt(-25)
            | full_train_csv["locationy"].gt(25)
        ).sum()
    })

range_check_summary = pd.DataFrame(range_checks)

print("\nRange-check results:")
display(range_check_summary)


# ============================================================
# 10. Logical consistency checks
# ============================================================

consistency_checks = []

if {"contested", "num_contesters"}.issubset(full_train_csv.columns):
    contested_numeric = (
        full_train_csv["contested"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "true": True,
            "false": False,
            "1": True,
            "0": False
        })
    )

    consistency_checks.append({
        "check": "contested=True but num_contesters=0",
        "violation_count": (
            contested_numeric.eq(True)
            & full_train_csv["num_contesters"].eq(0)
        ).sum()
    })

    consistency_checks.append({
        "check": "contested=False but num_contesters>0",
        "violation_count": (
            contested_numeric.eq(False)
            & full_train_csv["num_contesters"].gt(0)
        ).sum()
    })

if {"three", "distance"}.issubset(full_train_csv.columns):
    three_numeric = (
        full_train_csv["three"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "true": True,
            "false": False,
            "1": True,
            "0": False
        })
    )


    consistency_checks.append({
        "check": "three=True with distance < 20 feet",
        "violation_count": (
            three_numeric.eq(True)
            & full_train_csv["distance"].lt(20)
        ).sum()
    })

if consistency_checks:
    print("\nLogical-consistency checks:")
    display(pd.DataFrame(consistency_checks))

# ============================================================
# exploring out of boudns records for full_train_csv
# ============================================================
print("\n--- Exploring out of bounds records ---")

x_out_of_bounds = ~full_train_csv["locationx"].between(
    -47,
    47,
    inclusive="both"
)

y_out_of_bounds = ~full_train_csv["locationy"].between(
    -25,
    25,
    inclusive="both"
)

print(
    f"locationx outside [-47, 47]: "
    f"{x_out_of_bounds.sum():,} "
    f"({x_out_of_bounds.mean():.3%})"
)

print(
    f"locationy outside [-25, 25]: "
    f"{y_out_of_bounds.sum():,} "
    f"({y_out_of_bounds.mean():.3%})"
)

print(
    f"Outside either nominal boundary: "
    f"{(x_out_of_bounds | y_out_of_bounds).sum():,}"
)

location_problem_columns = [
    "shot_id",
    "outcome",
    "shottype",
    "distance",
    "locationx",
    "locationy",
    "three"
]

display(
    full_train_csv.loc[
        x_out_of_bounds | y_out_of_bounds,
        location_problem_columns
    ]
    .sort_values(["locationx", "locationy"])
    .head(50)
)

out_of_bounds_summary = pd.DataFrame({
    "locationx_below_-47": [
        full_train_csv["locationx"].lt(-47).sum()
    ],
    "locationx_above_47": [
        full_train_csv["locationx"].gt(47).sum()
    ],
    "locationy_below_-25": [
        full_train_csv["locationy"].lt(-25).sum()
    ],
    "locationy_above_25": [
        full_train_csv["locationy"].gt(25).sum()
    ]
})

display(out_of_bounds_summary)


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=full_train_csv.sample(
        n=min(50_000, len(full_train_csv)),
        random_state=42
    ),
    x="locationx",
    y="locationy",
    alpha=0.15,
    s=10
)

plt.axvline(-47, color="red", linestyle="--", label="Nominal baselines")
plt.axvline(47, color="red", linestyle="--")
plt.axhline(-25, color="orange", linestyle="--", label="Nominal sidelines")
plt.axhline(25, color="orange", linestyle="--")

plt.xlabel("locationx")
plt.ylabel("locationy")
plt.title("Shot Locations and Nominal Court Boundaries")
plt.legend()
plt.show()

# ============================================================
# exploring out of boudns records FOR THE TEST DATA
# ============================================================
print("\n--- Exploring out of bounds records in test data ---")

x_out_of_bounds = ~full_test_csv["locationx"].between(
    -47,
    47,
    inclusive="both"
)

y_out_of_bounds = ~full_test_csv["locationy"].between(
    -25,
    25,
    inclusive="both"
)

print(
    f"locationx outside [-47, 47]: "
    f"{x_out_of_bounds.sum():,} "
    f"({x_out_of_bounds.mean():.3%})"
)

print(
    f"locationy outside [-25, 25]: "
    f"{y_out_of_bounds.sum():,} "
    f"({y_out_of_bounds.mean():.3%})"
)

print(
    f"Outside either nominal boundary: "
    f"{(x_out_of_bounds | y_out_of_bounds).sum():,}"
)

location_problem_columns = [
    "shot_id",
    "shottype",
    "distance",
    "locationx",
    "locationy",
    "three"
]

display(
    full_test_csv.loc[
        x_out_of_bounds | y_out_of_bounds,
        location_problem_columns
    ]
    .sort_values(["locationx", "locationy"])
    .head(50)
)

out_of_bounds_summary = pd.DataFrame({
    "locationx_below_-47": [
        full_test_csv["locationx"].lt(-47).sum()
    ],
    "locationx_above_47": [
        full_test_csv["locationx"].gt(47).sum()
    ],
    "locationy_below_-25": [
        full_test_csv["locationy"].lt(-25).sum()
    ],
    "locationy_above_25": [
        full_test_csv["locationy"].gt(25).sum()
    ]
})

display(out_of_bounds_summary)


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=full_test_csv.sample(
        n=min(50_000, len(full_test_csv)),
        random_state=42
    ),
    x="locationx",
    y="locationy",
    alpha=0.15,
    s=10
)

plt.axvline(-47, color="red", linestyle="--", label="Nominal baselines")
plt.axvline(47, color="red", linestyle="--")
plt.axhline(-25, color="orange", linestyle="--", label="Nominal sidelines")
plt.axhline(25, color="orange", linestyle="--")

plt.xlabel("locationx")
plt.ylabel("locationy")
plt.title("Shot Locations and Nominal Court Boundaries for Test Data")
plt.legend()
plt.show()


# ============================================================
# derive features from closestdefapproach_parsed
# ============================================================

full_train_csv[
    [
        "closest_defender_distance_1_sec",
        "closest_defender_distance_0.75_sec",
        "closest_defender_distance_0.5_sec",
        "closest_defender_distance_0.25_sec",
    ]
] = pd.DataFrame(
    full_train_csv["closestdefapproach_parsed"].tolist(),
    index=full_train_csv.index
)

full_train_csv["closest_defender_dist_covered"] = full_train_csv["closest_defender_distance_1_sec"] - full_train_csv["closest_defender_distance_0.25_sec"]

full_train_csv["closest_defender_avg_speed"] = full_train_csv["closest_defender_dist_covered"] / 0.75

full_train_csv["recent_closing_amount"] = (
    full_train_csv["closest_defender_distance_0.5_sec"]
    - full_train_csv["closest_defender_distance_0.25_sec"]
)

stationary_cutoff = 1.0
rapid_cutoff = 6.0  

full_train_csv["closest_defender_motion"] = np.select(
    [
        full_train_csv["closest_defender_avg_speed"].abs() < stationary_cutoff,
        full_train_csv["closest_defender_avg_speed"] >= rapid_cutoff,
        full_train_csv["closest_defender_avg_speed"] <= -1*stationary_cutoff,
    ],
    [
        "stationary",
        "rapidly_closing",
        "moving_away",
    ],
    default="moderately_moving"
)



# ============================================================
#  Numeric-variable summaries
# ============================================================

numeric_columns = full_train_csv.select_dtypes(
    include=np.number
).columns

print("\nNumeric-variable summary:")
display(
    full_train_csv[numeric_columns]
        .describe()
        .T
)

# ============================================================
#  shot clock, shooter speed
# ============================================================
full_train_csv["shotclock_bin"] = pd.cut(
    full_train_csv["shotclock"],
    bins=[0, 2, 5, 10, 15, 20, 24],
    labels=["0-2", "2-5", "5-10", "10-15", "15-20", "20-24"],
    include_lowest=True
)
full_train_csv["distance_bin"] = pd.cut(
    full_train_csv["distance"],
    bins=[0, 5, 10, 20, np.inf],
    labels=["0–5 ft", "5–10 ft", "10–20 ft", "20+ ft"],
    include_lowest=True
)

full_train_csv["late_clock"] = full_train_csv["shotclock"] <= 2

#######################
full_train_csv["shooterspeed_bin"] = pd.qcut(
    full_train_csv["shooterspeed"],
    q=5,
    duplicates="drop"
)

# ============================================================
# basic shape and column information
# ============================================================

print("Dataset shape:", full_train_csv.shape)
print("\nColumn names:")
print(full_train_csv.columns.tolist())

print("\nFirst five rows:")
display(full_train_csv.head())


# ============================================================
# data types
# ============================================================

print("\nData types:")
display(full_train_csv.dtypes.rename("dtype").to_frame())
```

```python
def build_features(df):
    df = df.copy()

    # --------------------------------------------------------
    # Shot geometry
    # --------------------------------------------------------

    df["abs_locationy"] = df["locationy"].abs()

    rim_dx = df["locationx"] + 41.75
    rim_dy = df["locationy"]

    df["rim_angle"] = np.arctan2(
        np.abs(rim_dy),
        rim_dx
    )

    # --------------------------------------------------------
    # Contester-distance aggregates
    # --------------------------------------------------------

    contester_distance_columns = [
        "distcont1",
        "distcont2",
        "distcont3",
        "distcont4"
    ]

    df["min_contester_dist"] = df[
        contester_distance_columns
    ].min(axis=1, skipna=True)

    df["mean_contester_dist"] = df[
        contester_distance_columns
    ].mean(axis=1, skipna=True)

    df["max_contester_dist"] = df[
        contester_distance_columns
    ].max(axis=1, skipna=True)

    # --------------------------------------------------------
    # Shot-clock indicators
    # --------------------------------------------------------

    df["late_clock"] = (df["shotclock"] <= 2).astype(int)
    df["very_late_clock"] = (df["shotclock"] <= 1).astype(int)

    # --------------------------------------------------------
    # Dribble indicators
    # --------------------------------------------------------

    df["zero_dribble"] = (
        df["dribblesbefore"] == 0
    ).astype(int)

    # Preserve the original for CatBoost but add a capped
    # version so rare extreme values cannot dominate splits.
    df["dribbles_capped"] = df["dribblesbefore"].clip(
        upper=15
    )

    return df


def create_cumulative_feature_sets(feature_groups):
    cumulative_sets = OrderedDict()
    current_features = []

    for group_name, group_features in feature_groups.items():
        current_features = current_features + group_features
        cumulative_sets[group_name] = current_features.copy()

    return cumulative_sets

def prepare_catboost_data(df, features):
    X = df[features].copy()

    categorical_features = [
        feature
        for feature in features
        if feature in CATEGORICAL_COLUMNS
    ]

    for column in categorical_features:
        X[column] = (
            X[column]
            .fillna("MISSING")
            .astype(str)
        )

    return X, categorical_features
```

```python

from collections import OrderedDict


FEATURE_GROUPS = OrderedDict({
    # Group A: intrinsic shot difficulty
    "A_shot_geometry": [
        "shottype",
        "distance",
        "locationx",
        "locationy",
        "abs_locationy",
        "rim_angle",
        "three"
    ],

    # group b: defensive pressure at/near release
    "B_defensive_pressure": [
        "closestdefdist",
        "contested",
        "num_contesters",
        "distcont1",
        "distcont2",
        "min_contester_dist",
        "mean_contester_dist",
        "max_contester_dist"
    ],

    # Group c: defender trajectory before releaseee
    "C_defender_motion": [
        "closest_defender_distance_1_sec",
        "closest_defender_distance_0.75_sec",
        "closest_defender_distance_0.5_sec",
        "closest_defender_distance_0.25_sec",
        "closest_defender_dist_covered",
        "closest_defender_avg_speed",
        "recent_closing_amount"
    ],

    # Group D: possession and shot-creation context
    "D_play_context": [
        "shotclock",
        "late_clock",
        "very_late_clock",
        "dribblesbefore",
        "dribbles_capped",
        "zero_dribble",
        "shooterspeed",
        "gamestate"
    ],

    # Group E: shooter skill
    "E_shooter_identity": [
        "shooter_id"
    ],

    # Group F: defender identity
    "F_defender_identity": [
        "closestdef_id",
        "contester1_id"
    ],

    # Group G: team and temporal context
    "G_team_time": [
        "off_team_id",
        "def_team_id",
        "season_id",
        "month"
    ]
})

CUMULATIVE_FEATURE_SETS = create_cumulative_feature_sets(
    FEATURE_GROUPS
)

for model_name, features in CUMULATIVE_FEATURE_SETS.items():
    print(model_name, len(features))

CATEGORICAL_COLUMNS = {
    "shottype",
    "gamestate",
    "shooter_id",
    "closestdef_id",
    "contester1_id",
    "off_team_id",
    "def_team_id",
    "season_id",
    "month"
}
```

```python
from sklearn.model_selection import train_test_split

screen_train_indices, screen_tune_indices = train_test_split(
    np.arange(len(development_features)),
    test_size=0.20,
    random_state=42,
    stratify=y_development
)

SCREENING_PARAMS = {
    "loss_function": "Logloss",
    "eval_metric": "Logloss",
    "iterations": 1500,
    "learning_rate": 0.05,
    "depth": 7,
    "l2_leaf_reg": 8,
    "random_seed": 42,
    "verbose": False,
    "allow_writing_files": False,
    "thread_count": -1
}

from catboost import CatBoostClassifier
from sklearn.metrics import log_loss


def run_ablation_screening(
    development_data,
    target,
    feature_sets,
    train_indices,
    tune_indices,
    model_params
):
    results = []
    fitted_models = {}

    for model_name, features in feature_sets.items():
        print(f"Training {model_name}...")

        X, categorical_features = prepare_catboost_data(
            development_data,
            features
        )

        X_train = X.iloc[train_indices]
        y_train = target.iloc[train_indices]

        X_tune = X.iloc[tune_indices]
        y_tune = target.iloc[tune_indices]

        model = CatBoostClassifier(**model_params)

        model.fit(
            X_train,
            y_train,
            cat_features=categorical_features,
            eval_set=(X_tune, y_tune),
            early_stopping_rounds=100,
            use_best_model=True
        )

        tune_probabilities = model.predict_proba(
            X_tune
        )[:, 1]

        tuning_logloss = log_loss(
            y_tune,
            tune_probabilities
        )

        results.append({
            "feature_set": model_name,
            "num_features": len(features),
            "tuning_logloss": tuning_logloss,
            "best_iteration": model.get_best_iteration()
        })

        fitted_models[model_name] = model

    results_df = (
        pd.DataFrame(results)
        .sort_values("tuning_logloss")
        .reset_index(drop=True)
    )

    return results_df, fitted_models


ablation_results, ablation_models = run_ablation_screening(
    development_data=development_features,
    target=y_development,
    feature_sets=CUMULATIVE_FEATURE_SETS,
    train_indices=screen_train_indices,
    tune_indices=screen_tune_indices,
    model_params=SCREENING_PARAMS
)

display(ablation_results)


ablation_results_ordered = (
    ablation_results
    .set_index("feature_set")
    .loc[list(CUMULATIVE_FEATURE_SETS.keys())]
    .reset_index()x
)

ablation_results_ordered["improvement_from_previous"] = (
    ablation_results_ordered["tuning_logloss"]
    .shift(1)
    - ablation_results_ordered["tuning_logloss"]
)

display(ablation_results_ordered)
```

```python
import gc
import numpy as np
import pandas as pd

from catboost import CatBoostClassifier
from sklearn.metrics import log_loss
from sklearn.model_selection import StratifiedKFold

FINALIST_FEATURE_SETS = [
    "E_shooter_identity",
    "F_defender_identity",
    "G_team_time"
]

CV_PARAMS = {
    "loss_function": "Logloss",
    "eval_metric": "Logloss",

   
    "iterations": 1800,

    "learning_rate": 0.05,
    "depth": 7,
    "l2_leaf_reg": 8,

    "random_seed": 42,
    "thread_count": -1,
    "allow_writing_files": False,
    "verbose": False
}

N_SPLITS = 3
RANDOM_STATE = 42
EARLY_STOPPING_ROUNDS = 100

development_features_cv = (
    development_features
    .reset_index(drop=True)
    .copy()
)

y_development_cv = (
    y_development
    .reset_index(drop=True)
    .astype(int)
)

cv_splitter = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_splits = list(
    cv_splitter.split(
        development_features_cv,
        y_development_cv
    )
)

for fold_number, (train_idx, fold_idx) in enumerate(
    cv_splits,
    start=1
):
    print(
        f"Fold {fold_number}: "
        f"{len(train_idx):,} training rows, "
        f"{len(fold_idx):,} evaluation rows"
    )


def cross_validate_feature_set(
    data,
    target,
    feature_set_name,
    features,
    cv_splits,
    model_params,
    early_stopping_rounds=100
):
    """
    Run CatBoost cross-validation for one feature set.

    Returns:
        fold_results:
            One row per fold with log-loss and iteration count.

        summary:
            Aggregate CV results.

        oof_probabilities:
            Out-of-fold probability for every development row.
    """

    X, categorical_features = prepare_catboost_data(
        data,
        features
    )

    target = target.reset_index(drop=True)

    oof_probabilities = np.full(
        len(X),
        np.nan,
        dtype=float
    )

    fold_records = []

    for fold_number, (train_idx, fold_idx) in enumerate(
        cv_splits,
        start=1
    ):
        print(
            f"{feature_set_name}: "
            f"fold {fold_number}/{len(cv_splits)}"
        )

        X_train = X.iloc[train_idx]
        y_train = target.iloc[train_idx]

        X_fold = X.iloc[fold_idx]
        y_fold = target.iloc[fold_idx]

        model = CatBoostClassifier(**model_params)

        model.fit(
            X_train,
            y_train,
            cat_features=categorical_features,
            eval_set=(X_fold, y_fold),
            early_stopping_rounds=early_stopping_rounds,
            use_best_model=True,
            verbose=False
        )

        fold_probabilities = model.predict_proba(
            X_fold
        )[:, 1]

        oof_probabilities[fold_idx] = fold_probabilities

        fold_logloss = log_loss(
            y_fold,
            fold_probabilities
        )

        # CatBoost returns a zero-based tree index.
        best_iteration_index = model.get_best_iteration()
        best_tree_count = best_iteration_index + 1

        fold_records.append({
            "feature_set": feature_set_name,
            "fold": fold_number,
            "train_rows": len(train_idx),
            "evaluation_rows": len(fold_idx),
            "logloss": fold_logloss,
            "best_iteration_index": best_iteration_index,
            "best_tree_count": best_tree_count
        })

        del model
        gc.collect()

    if np.isnan(oof_probabilities).any():
        raise ValueError(
            f"Some OOF predictions are missing for "
            f"{feature_set_name}."
        )

    fold_results = pd.DataFrame(fold_records)

    summary = {
        "feature_set": feature_set_name,
        "num_features": len(features),
        "mean_fold_logloss": fold_results[
            "logloss"
        ].mean(),
        "std_fold_logloss": fold_results[
            "logloss"
        ].std(ddof=1),
        "oof_logloss": log_loss(
            target,
            oof_probabilities
        ),
        "median_best_tree_count": int(
            np.median(
                fold_results["best_tree_count"]
            )
        ),
        "min_best_tree_count": int(
            fold_results["best_tree_count"].min()
        ),
        "max_best_tree_count": int(
            fold_results["best_tree_count"].max()
        )
    }

    return {
        "fold_results": fold_results,
        "summary": summary,
        "oof_probabilities": oof_probabilities
    }

cv_results = {}

for feature_set_name in FINALIST_FEATURE_SETS:
    print(
        "\n"
        + "=" * 60
        + f"\nCross-validating {feature_set_name}\n"
        + "=" * 60
    )

    feature_columns = CUMULATIVE_FEATURE_SETS[
        feature_set_name
    ]

    result = cross_validate_feature_set(
        data=development_features_cv,
        target=y_development_cv,
        feature_set_name=feature_set_name,
        features=feature_columns,
        cv_splits=cv_splits,
        model_params=CV_PARAMS,
        early_stopping_rounds=EARLY_STOPPING_ROUNDS
    )

    cv_results[feature_set_name] = result


    cv_summary = pd.DataFrame([
    result["summary"]
    for result in cv_results.values()
]).sort_values(
    "oof_logloss"
).reset_index(drop=True)

display(cv_summary)
```

```python
all_fold_results = pd.concat(
    [
        result["fold_results"]
        for result in cv_results.values()
    ],
    ignore_index=True
)

fold_logloss_comparison = (
    all_fold_results
    .pivot(
        index="fold",
        columns="feature_set",
        values="logloss"
    )
)

display(fold_logloss_comparison)

fold_logloss_comparison[
    "improvement_E_to_F"
] = (
    fold_logloss_comparison["E_shooter_identity"]
    - fold_logloss_comparison["F_defender_identity"]
)

fold_logloss_comparison[
    "improvement_F_to_G"
] = (
    fold_logloss_comparison["F_defender_identity"]
    - fold_logloss_comparison["G_team_time"]
)

display(fold_logloss_comparison)

fold_iteration_comparison = (
    all_fold_results
    .pivot(
        index="fold",
        columns="feature_set",
        values="best_tree_count"
    )
)

display(fold_iteration_comparison)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(
    data=all_fold_results,
    x="feature_set",
    y="logloss",
    errorbar="sd",
    color="steelblue"
)

plt.ylabel("Fold Log-Loss")
plt.xlabel("Feature Set")
plt.title("Three-Fold Cross-Validation by Feature Set")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))

sns.lineplot(
    data=all_fold_results,
    x="feature_set",
    y="logloss",
    hue="fold",
    marker="o",
    palette="tab10"
)

plt.ylabel("Log-Loss")
plt.xlabel("Feature Set")
plt.title("Paired Fold Performance Across Finalists")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

selected_feature_set = cv_summary.iloc[0][
    "feature_set"
]

selected_tree_count = int(
    cv_summary.iloc[0]["median_best_tree_count"]
)

print("Selected feature set:", selected_feature_set)
print("Selected number of trees:", selected_tree_count)

final_params = CV_PARAMS.copy()
final_params["iterations"] = selected_tree_count
```
